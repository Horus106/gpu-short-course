## Example 1

```python
import numpy as np


# Generate an array of 20 random integers between -1 and 1
x = np.linspace(-1, 1, 20)

print(x)
```

```python
import matplotlib.pyplot as plt

f_x = x**2
g_x = 2 * x

plt.plot(x, f_x, label='f(x) = x^2')
plt.plot(x, g_x, label='g(x) = 2x')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Plot of f(x) and g(x)')
plt.legend()
plt.grid()
plt.show()
```
